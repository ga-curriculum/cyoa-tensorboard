```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPool2D, Flatten, Dense, Dropout
from tensorflow.keras.datasets import mnist
from tensorflow.keras.callbacks import TensorBoard
```

```python
%load_ext tensorboard
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()

X_train = (X_train / 255).reshape(-1, 28, 28, 1)
X_test = (X_test / 255).reshape(-1, 28, 28, 1)
```

```python
# Define the neural network model
model = Sequential()
model.add(Conv2D(64, 3, activation='relu', input_shape=(28, 28, 1)))
model.add(MaxPool2D(2))
model.add(Flatten())
model.add(Dense(32, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(10, activation='softmax'))

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy',  metrics=['acc'])
```

```python
# Create a TensorBoard callback
log_dir = "logs/fit2"  # Directory where the logs will be stored
tb = TensorBoard(log_dir=log_dir, histogram_freq=1)

# Train the model and log data to TensorBoard
#  model.fit(x_train, y_train, epochs=5, callbacks=[tensorboard_callback], validation_data=(x_test, y_test))

model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    batch_size=256,
    epochs=2,
    callbacks=[tb]
)
```
